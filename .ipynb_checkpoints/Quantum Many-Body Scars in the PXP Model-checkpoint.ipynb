Import libraries

# Quantum Many-Body Scars in the PXP Model

## Introduction

Most interacting quantum systems quickly relax or thermalize so that local measurements take steady values. An exception is the $PXP$ model of a Rydberg atom chain with a blockade constraint. For certain product states, unitary dynamics show large revivals and the system repeatedly returns close to its initial configuration. This many body scarring is revealed by the return probability, found by repeatedly measuring all qubits in the $Z$ basis and counting how often the outcome matches the starting bitstring.

In this lab, the time evolution of an open $PXP$ chain is simulated, and scarring is demonstrated by plotting the return probability of selected $Z$ basis product states over time.

### The model

The PXP model describes a Rydberg atom chain with a blockade constraint. With $|1\rangle$ as an excited Rydberg atom and $|0\rangle$ as the ground state, the projector onto $|0\rangle$ on site $i$ is
$$
P_i = |0\rangle\langle 0|_i = \frac{I + Z_i}{2},
$$
since $\frac{I+Z_i}{2}|0\rangle = \frac{|0\rangle + |0\rangle}{2} = |0\rangle$ and $\frac{I+Z_i}{2}|1\rangle = \frac{|1\rangle - |1\rangle}{2} = 0$.

The transverse drive on site $i$ is $X_i$, with strength $g$. For open boundaries, the PXP Hamiltonian is
$$
H_{\text{PXP}} = g \sum_{i=2}^{L-1} P_{i-1} X_i P_{i+1}.
$$
The flip on site $i$ is projected by the neighboring projectors $P_{i-1}$ and $P_{i+1}$. If either neighbor is in $|1\rangle$, the corresponding projector annihilates the state and $X_i$ has no effect. A spin can only flip to $|1\rangle$ when both neighbors are $|0\rangle$, so configurations with adjacent $|1\rangle$ states are never produced, which enforces the Rydberg blockade (provided the initial state has no adjacent $|1\rangle$ spins).

## A

To implement H, first expand the local projector structure into a sum of Pauli strings. For one term, $P_{i-1} X_i P_{i+1}= \frac{1}{2}(I + Z_{i-1})\, X_i\, \frac{1}{2}(I + Z_{i+1})= \frac{1}{4}\left( X_i + Z_{i-1} X_i + X_i Z_{i+1} + Z_{i-1} X_i Z_{i+1} \right).$ So with $g = 1$ and $L = 10$, the Hamiltonian as a Pauli string sum is $H = \frac{1}{4} \sum_{i=2}^{9} \left(X_i + Z_{i-1} X_i + X_i Z_{i+1} + Z_{i-1} X_i Z_{i+1}\right)$, where each operator is tensored with identities on all other qubits.

```python
def pxp_hamiltonian(L= 10, g= 1):
    coef = 0.25 * g
    sparse_terms = []

    #i = 1 ... L-2 (qubit 0 and L-1 are boundaries)
    for i in range(1, L - 1):
        sparse_terms.append(("X",[i],coef))
        
        sparse_terms.append(("ZX",[i - 1, i],coef))
        
        sparse_terms.append(("XZ",[i, i + 1],coef))
        
        sparse_terms.append(("ZXZ",[i - 1, i, i + 1],coef))

    H = SparsePauliOp.from_sparse_list(sparse_terms, num_qubits=L)
    H = H.simplify()
    return H

#build H
H = pxp_hamiltonian()
```

Expect $10$ qubits and $4\times8=32$ Pauli terms.

```python
print("num_qubits =", H.num_qubits)
print("number of Pauli terms =", H.size)
```

## B

To prepare the initial product states, a simple circuit is constructed on 10 qubits. Starting from $|0\cdots 0\rangle$, the Néel state $|Z_2\rangle = |1010101010\rangle$ is obtained by applying an X gate to qubits 1,3,5,7,9, while the zero state $|0000000000\rangle$ is left unchanged.

```python
def neel_circuit(L= 10):
    qc = QuantumCircuit(L, name="Neel")
    for q in range(1, L, 2):
        qc.x(q)# flip those sites to 1
    return qc

def zero_circuit(L=10):
    return QuantumCircuit(L, name="Zero")

#for extension
def cluster_circuit(L=10):
    qc = QuantumCircuit(L, name="Cluster")
    for q in range(L//2, L):
        qc.x(q)
    return qc
# this is a random circuit I thought of with no specific pattern, so will not generalise it!
def half_circuit(L=10):
    bitstring = "0100110011"      
    qc = QuantumCircuit(L, name="half")
    for q in range(L):
        if bitstring[L - 1 - q] == "1":
            qc.x(q)
    return qc

def block_circuit(L=10):
    bitstring = "1001001001"      
    qc = QuantumCircuit(L, name="block")
    for q in range(L):
        if bitstring[L - 1 - q] == "1":
            qc.x(q)
    return qc

# Circuits
qc_neel = neel_circuit()
qc_zero = zero_circuit()
```

## C

For the digital time evolution, a product formula circuit for $U(T)$ is built from $r = T / \delta$ identical steps. The time evolution for a total time $T = 20$ with step sizes $\delta = 0.1$ and $\delta = 0.2$ is shown below, using second order Suzuki-Trotter.

```python
def evolution_circuit(delta,order, L= 10, T= 20,H=H,):

    r = round(T / delta)
    if order == 1:
        synth = LieTrotter()
    elif order == 2:
        synth = SuzukiTrotter(order=2)

    # one step evolution gate
    single_step_gate = PauliEvolutionGate(H, time=delta, synthesis=synth)

    # single step as a circuit
    single_step_evolution = QuantumCircuit(L, name=f"o{order}_dt{delta}")
    single_step_evolution.append(single_step_gate, single_step_evolution.qubits)

    # full evolution U(T) as r repeated steps
    evolved_state = QuantumCircuit(L, name=f"U_T_dt{delta}_o{order}")
    for _ in range(r):
        evolved_state.append(single_step_evolution, evolved_state.qubits)

    return evolved_state
delta_02 = 0.2 #100 steps
delta_01 = 0.1 #200 steps

U_T_02 = evolution_circuit(delta=delta_02,order=2) 
U_T_01 = evolution_circuit(delta=delta_01,order=2) 
```

For $\delta=0.1$, the single step time evolution circuit is shown below. The circuit for $\delta=0.2$ has the same structure, but with rescaled rotation angles.

```python
single_step_gate = PauliEvolutionGate(H, time=0.1, synthesis=SuzukiTrotter(order=2))


single_step_evolution = QuantumCircuit(10, name="single_step")
single_step_evolution.append(single_step_gate, single_step_evolution.qubits)
print(
    f"""
Depth: {single_step_evolution.decompose(reps=3).depth()}
Gate count: {len(single_step_evolution.decompose(reps=3))}
Nonlocal gate count: {single_step_evolution.decompose(reps=3).num_nonlocal_gates()}
Gate breakdown: {", ".join([f"{k.upper()}: {v}" for k, v in single_step_evolution.decompose(reps=3).count_ops().items()])}
"""
)
single_step_evolution.decompose(reps=2).draw("mpl", fold=-1)
```

## D

To sample the dynamics, a separate circuit is built for each time step $k = 0,1,\dots,r$. Each circuit prepares the chosen initial product state, applies $k$ Trotter steps (total time $t_k = k\delta$), and then measures all qubits in the $Z$ basis. Using $8192$ shots on the AerSimulator, the return probability at time $t_k$ is estimated as the fraction of shots that reproduce the exact initial bitstring, and these values are plotted as a function of time for both initial states and both Trotter orders.

```python
def kstep_circuit(L, H, delta, order, k, init_state):
    if init_state == "neel":
        init = neel_circuit(L)
    elif init_state == "Zero":
        init = zero_circuit(L)
    # extension
    elif init_state == "cluster":
        init = cluster_circuit(L)
    elif init_state == "half":
        init = half_circuit(L)
    elif init_state == "block":
        init = block_circuit(L)


    
    qc = QuantumCircuit(L, name=f"{init_state}_o{order}_k{k}")
    qc.compose(init, inplace=True)

    # Evolution for time T = k * delta
    evol = evolution_circuit(delta=delta, L=L, T=k * delta, H=H, order=order)
    qc.compose(evol, inplace=True)

    qc.measure_all()  # Z measurement on all qubits
    return qc


def full_evolution(delta,order, init_state, L=10, T=20, H=H, shots = 8192, seed = 1234):
    r = round(T / delta)

    circuits = [kstep_circuit(L, H, delta, order, k, init_state) for k in range(r + 1)]

    backend = AerSimulator(seed_simulator=seed)
    trancirc = transpile(circuits, backend)
    job = backend.run(trancirc, shots=shots)
    result = job.result()


    if init_state == "neel":
        qstring = "".join("1" if (i % 2 == 0) else "0" for i in range(L)) #"101..."
    elif init_state == "Zero":
        qstring= "0" * L #"000..."
    # extension
    elif init_state == "cluster":
        qstring = "1" * (L//2) + "0" * (L//2)# "111...000"
    elif init_state == "half":
        qstring = "0100110011"
    elif init_state == "block":
        qstring = "1001001001"

    times = np.arange(r + 1) * delta
    probs = np.zeros(r + 1)

    for k, qc in enumerate(circuits):
        counts = result.get_counts(qc)
        occ = counts.get(qstring, 0)# how often is the initial string reproduced
        probs[k] = occ / shots

    return dict(times=times, probs=probs, inputs=dict(L=L, T=T, delta=delta, order=order, init=init_state))

# part e
def top_two_peaks(times, probs):
    times = np.asarray(times)
    probs = np.asarray(probs)

    peak_indices = []

    # find local maxima
    for i in range(1, len(probs) - 1):
        if probs[i] >= probs[i - 1] and probs[i] >= probs[i + 1] and times[i] > 0:
            peak_indices.append(i)

    height_index_pairs = []
    for i in peak_indices:
        height_index_pairs.append((probs[i], i))
    # largest first
    height_index_pairs.sort(reverse=True)
    top_pairs = height_index_pairs[:2]

    peaks = []
    for p, i in top_pairs:
        t = times[i]
        peaks.append((t,p))

    return peaks


for delta in [0.2, 0.1]:

    values = []
    for init_state in ["neel", "Zero"]:
        for order in [1, 2]:
            val = full_evolution(delta=delta,order=order, init_state=init_state)
            values.append(val)

    for val in values:
        #e
        peaks = top_two_peaks(val["times"], val["probs"])
        print(f"delta={delta}, {val['inputs']['init']} order {val['inputs']['order']} peaks: {peaks}")
        #plot
        plt.plot(val['times'], val['probs'], label=f"{val['inputs']['init']} Order {val['inputs']['order']}")

    plt.xlabel("time t")
    plt.ylabel("return probability")
    plt.title(f"PXP return probability for L=10, delta={delta}")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

## e

Looking at the plot for $\delta = 0.1$, for the Néel initial state clear many body scars are visible. For Trotter order 1, the first two prominent revivals occur at $t \approx 4.7$ with height $\approx 0.878$ and at $t \approx 9.3$ with height $\approx 0.689$. For order 2, the first revival is at $t \approx 4.7$ with height $\approx 0.888$, and the second again at $t \approx 9.3$ with height $\approx 0.716$. In contrast, the generic zero product state shows no large revivals. The return probability only reaches about $0.331$ (order 1) or $0.366$ (order 2), at $t \approx 18.4$, with broad low bumps. Roughly the same behaviour is seen for $\delta = 0.2$, with lower resolution in time but the scarred dynamics still visible.

These results are consistent with the literature on the PXP model, which finds a small set of special “scarred” eigenstates that overlap strongly with the $|Z_2\rangle$ Néel product state and lie at almost regularly spaced energies inside an otherwise thermal spectrum. When the system is started in $|Z_2\rangle$, its evolution is dominated by these special states and strong revivals appear in time, whereas generic product states spread over many thermal eigenstates and relax quickly without large revivals [1].

## Extension

Following from the theory, it is clear that the strong revivals are expected to be specific to the $|Z_2\rangle$ Néel product state, and generic states should not show this behavior. This was shown above when comparing to the generic zero product state, but what happens for a more clustered state such as $|1111100000\rangle$? The initial states considered in the extension do not strictly respect the Rydberg blockade constraint (they contain adjacent $|1\rangle$ spins), but the results are still interesting.

```python
for delta in [0.2, 0.1]:

    values = []
    for init_state in ["neel", "cluster"]:
        for order in [1, 2]:
            val = full_evolution(delta=delta,order=order, init_state=init_state)
            values.append(val)

    for val in values:
        #e
        peaks = top_two_peaks(val["times"], val["probs"])
        print(f"delta={delta}, {val['inputs']['init']} order {val['inputs']['order']} peaks: {peaks}")
        #plot
        plt.plot(val['times'], val['probs'], label=f"{val['inputs']['init']} Order {val['inputs']['order']}")

    plt.xlabel("time t")
    plt.ylabel("return probability")
    plt.title(f"PXP return probability for L=10, delta={delta}")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

This is not what I expected to see. To understand why this happens, the clustered state $|1111100000\rangle$ can be decomposed in the eigenbasis of the PXP Hamiltonian. To do this, the full Hamiltonian matrix $H$ is first constructed and diagonalized to find its eigenvectors $|E_n\rangle$. The clustered product state is then written as a statevector $|C\rangle$ and expanded in this eigenbasis by $c_n = \langle E_n | C\rangle$. The weights $|c_n|^2$ are computed, and the largest ones are listed to show which eigenstates of $H$ contribute most strongly to $|C\rangle$.

```python
#Full matrix of H
H_full = H.to_matrix() 

#diagonalize
evals, evecs = np.linalg.eigh(H_full)

# get statevector of |1111100000>
psi_C = Statevector.from_label("1111100000").data 

# c_n = <E_n|C>
c = np.conjugate(evecs).T @ psi_C 


weights = np.abs(c)**2

# print largest few weights
idx_sorted = np.argsort(weights)[::-1]
for k in range(5):
    n = idx_sorted[k]
    print(f"|c_{n}|^2 = {weights[n]:.4f}")
```

Only a small set of four eigenstates carry essentially all the weight in the decomposition of $|1111100000\rangle$, which explains why the return probability shows strong revivals. This structure arises because the relevant eigenstates lie in a smaller symmetry subspace, rather than from scarring as in the Néel case. Consider now the decomposition of the Néel product state $|Z_2\rangle = |1010101010\rangle$

```python
#Full matrix of H
H_full = H.to_matrix() 

#diagonalize
evals, evecs = np.linalg.eigh(H_full)

# get statevector of |1010101010>
psi_C = Statevector.from_label("1010101010").data 

#Expand |C> in the eigenbasis, c_n = <E_n|C>
c = np.conjugate(evecs).T @ psi_C 


weights = np.abs(c)**2

# print largest few weights
idx_sorted = np.argsort(weights)[::-1]
for k in range(10):
    n = idx_sorted[k]
    print(f"|c_{n}|^2 = {weights[n]:.4f}")
```

The expansion coefficients show a very similar structure. Most of the weight is concentrated in a small set of eigenstates, while the remaining eigenstates carry very little weight.

For the zero product state $|0000000000\rangle$, it is expected that the weights will be more evenly distributed.

```python
#Full matrix of H
H_full = H.to_matrix() 

#diagonalize
evals, evecs = np.linalg.eigh(H_full)

# get statevector of |0000000000>
psi_C = Statevector.from_label("0000000000").data 

#Expand |C> in the eigenbasis, c_n = <E_n|C>
c = np.conjugate(evecs).T @ psi_C 


weights = np.abs(c)**2

# print largest few weights
idx_sorted = np.argsort(weights)[::-1]
for k in range(20):
    n = idx_sorted[k]
    print(f"|c_{n}|^2 = {weights[n]:.4f}, eigenvalue = {evals[n]:.4f}")
```

As expected, the weights for the zero product state are spread over many eigenstates with comparable size, and the corresponding eigenvalues have many different values. This broad distribution in both weight and energy leads to only small, broad fluctuations in the return probability instead of sharp revivals.

From these examples, it seems that an input state with half of the spins in $|1\rangle$ and half in $|0\rangle$ may show signs of revivals. Now consider the state $|0100110011\rangle$.

```python
for delta in [0.2, 0.1]:

    values = []
    for init_state in ["neel", "half"]:
        for order in [1, 2]:
            val = full_evolution(delta=delta,order=order, init_state=init_state)
            values.append(val)

    for val in values:
        #e
        peaks = top_two_peaks(val["times"], val["probs"])
        print(f"delta={delta}, {val['inputs']['init']} order {val['inputs']['order']} peaks: {peaks}")
        #plot
        plt.plot(val['times'], val['probs'], label=f"{val['inputs']['init']} Order {val['inputs']['order']}")

    plt.xlabel("time t")
    plt.ylabel("return probability")
    plt.title(f"PXP return probability for L=10, delta={delta}")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

These show near perfect oscillations!

```python
#Full matrix of H
H_full = H.to_matrix() 

#diagonalize
evals, evecs = np.linalg.eigh(H_full)

# get statevector of |0100110011>
psi_C = Statevector.from_label("0100110011").data 

#Expand |C> in the eigenbasis, c_n = <E_n|C>
c = np.conjugate(evecs).T @ psi_C 


weights = np.abs(c)**2

# print largest few weights
idx_sorted = np.argsort(weights)[::-1]
for k in range(20):
    n = idx_sorted[k]
    print(f"|c_{n}|^2 = {weights[n]:.4f}, eigenvalue = {evals[n]:.4f}")
```

Interestingly, most of the weight is on eigenstates with energies very close to $0$ or $\pm \sqrt{2}$. This means that only a few frequencies contribute to the evolution, which explains the almost perfect oscillations. The mechanism seems slightly different than that of the state $|1111100000\rangle$, however the outcome is quite similar.

Moving to an initial state that satisfies the Rydberg blockade constraint, consider the state $|1001001001\rangle$. This state is not expected to show sharp revivals.

```python
for delta in [0.2, 0.1]:

    values = []
    for init_state in ["neel", "block"]:
        for order in [1, 2]:
            val = full_evolution(delta=delta,order=order, init_state=init_state)
            values.append(val)

    for val in values:
        #e
        peaks = top_two_peaks(val["times"], val["probs"])
        print(f"delta={delta}, {val['inputs']['init']} order {val['inputs']['order']} peaks: {peaks}")
        #plot
        plt.plot(val['times'], val['probs'], label=f"{val['inputs']['init']} Order {val['inputs']['order']}")

    plt.xlabel("time t")
    plt.ylabel("return probability")
    plt.title(f"PXP return probability for L=10, delta={delta}")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

This is close to what is expected, although the peak at $t\approx8.2$ with height $\approx0.727$ is somewhat stronger than expected. 

```python
#Full matrix of H
H_full = H.to_matrix() 

#diagonalize
evals, evecs = np.linalg.eigh(H_full)

# get statevector of |1001001001>
psi_C = Statevector.from_label("1001001001").data 

#Expand |C> in the eigenbasis, c_n = <E_n|C>
c = np.conjugate(evecs).T @ psi_C 


weights = np.abs(c)**2

# print largest few weights
idx_sorted = np.argsort(weights)[::-1]
for k in range(20):
    n = idx_sorted[k]
    print(f"|c_{n}|^2 = {weights[n]:.4f}, eigenvalue = {evals[n]:.4f}")
```

The largest contribution comes from a single pair of eigenstates at $E \approx \pm 0.77$ (with $\approx 44\%$ of the weight). This can explain the relatively large peak around $t \approx 8.2$, but the weight is more spread out than in the Néel or clustered cases, so the revivals are weaker and less regular.

## Conclusion

In this lab the PXP Hamiltonian for a ten site Rydberg chain was implemented as a Pauli string sum and used to approximate the time evolution with first and second order product formulas. Starting from the Néel state and a generic zero product state, the return probability was estimated from repeated $Z$ basis measurements using an Aer simulator. The Néel state shows clear many body scars, with large, sharp revivals that are seen more accurately with the second order Suzuki-Trotter expansion and the smaller step size. The zero product state instead shows only small, broad fluctuations in the return probability and no clear revivals, which matches the expectation that most product states spread over many thermal eigenstates and relax quickly.

In the extension, several other product states were tested. The clustered state $|1111100000\rangle$ and the state $|0100110011\rangle$ show very strong or even near perfect oscillations in the return probability, while the initial state that respects the blockade $|1001001001\rangle$ only shows a single moderate peak and then weaker, less regular behavior. By decomposing these initial states in the eigenbasis of the PXP Hamiltonian, it becomes clear that strong revivals arise when most of the weight sits on a small number of eigenstates with only a few distinct energy gaps, while more generic states spread their weight over many eigenstates with a wide range of energies. These observations show that even a small system already has dynamics that depend strongly on the structure of the initial product state.
