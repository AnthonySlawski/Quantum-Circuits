# Quantum Simulation

## Introduction

Trotterization approximates the unitary time evolution of a state under a time independent Hamiltonian. From the Schrödinger equation, a state initially in $|\psi(0)\rangle$ evolves as $|\psi(t)\rangle = e^{-iHt} |\psi(0)\rangle,$ where $H$ is the Hamiltonian. The Hamiltonians considered can be written as $H = \sum_j a_j P_j ,$ where $P_j$ are tensor products of Pauli matrices acting on n qubits, which may or may not commute.

For small time steps the evolution can be approximated as $e^{-i(H_1 + H_2)t}\approx e^{-iH_1 t} e^{-iH_2 t}$. Repeating this short time evolution $r$ times with $\Delta t = t/r$, gives the
first order Trotterization $|\psi(t)\rangle \approx\left( \prod_j e^{-i a_j P_j t / r} \right)^r |\psi(0)\rangle .$ Smaller $\Delta t$ improves the approximation but increases circuit depth.

The real time evolution of the transverse field Ising model on linear lattices with $N = 2$ and $N = 6$ sites are considered here. The spins $\sigma_i$ interact with nearest neighbors and feel an external field:
$$
H = -J \sum_{i=0}^{N-2} Z_i Z_{i+1}
    - h \sum_{i=0}^{N-1} X_i ,
$$
with $J$ Interaction energy, and $h$ the magnitude of an external field. More generally with a field at angle $\alpha$ in the $xz$ plane (going from $x$ to $z$ as $\alpha$ increases),
$$
H = -J \sum_{i=0}^{N-2} Z_i Z_{i+1}
    - h \sum_{i=0}^{N-1} \bigl( \sin\alpha\, Z_i + \cos\alpha\, X_i \bigr) .
$$

## Mapping the problem

The Hamiltonian above is mapped as shown below.

```python
def get_hamiltonian(nqubits, J, h, alpha):
    # List of Hamiltonian terms as 3-tuples containing
    # (1) the Pauli string,
    # (2) the qubit indices corresponding to the Pauli string,
    # (3) the coefficient.
    ZZ_tuples = [("ZZ", [i, i + 1], -J) for i in range(0, nqubits - 1)]
    Z_tuples = [("Z", [i], -h * np.sin(alpha)) for i in range(0, nqubits)]
    X_tuples = [("X", [i], -h * np.cos(alpha)) for i in range(0, nqubits)]
 
    # We create the Hamiltonian as a SparsePauliOp, via the method
    # `from_sparse_list`, and multiply by the interaction term.
    hamiltonian = SparsePauliOp.from_sparse_list(
        [*ZZ_tuples, *Z_tuples, *X_tuples], num_qubits=nqubits
    )
    return hamiltonian.simplify()
```

$ZZ_{\text{tuples}}$ creates nearest-neighbor couplings, for each pair $(i, i+1)$ it adds the Pauli string $ZZ$ acting on those two qubits with coefficient $-J$. Qubit $0$ is coupled to qubit $1$, qubit $1$ to qubit $2$, and so on up to qubit $n-2$ to qubit $n-1$, with no periodic connection.

Consider the system below with $N=6, J=0.2, h=1.2, \text{ and } \alpha=\frac{\pi}{8}.$

```python
n_qubits = 6
 
hamiltonian = get_hamiltonian(nqubits=n_qubits, J=0.2, h=1.2, alpha=np.pi / 8.0)
hamiltonian
```

Understanding the output, $IIIIZZ$ acts with the identity on qubits $2$ to $5$, and applies $ZZ$ to qubits $0$ and $1$, with coefficient $-J=-0.2$.


$IIIIIZ$ has coefficient $-h\sin\alpha=-1.2\sin\frac{\pi}{8}\approx-0.45922012$

$IIIIIX$ has coefficient $-h\cos\alpha=-1.2\cos\frac{\pi}{8}\approx-1.10865544$

## Three Trotterization techniques

Lie-Trotter (first order), second-order Suzuki-Trotter, and fourth-order Suzuki-Trotter time evolution techniques are used below. The techniques are explained in detail in the [Quantum Simulation notes](https://github.com/AnthonySlawski/Quantum-Circuits/blob/main/Quantum_Simulation_notes.pdf).

```python
num_timesteps = 60
evolution_time = 30.0
dt = evolution_time / num_timesteps
product_formula_lt = LieTrotter()
product_formula_st2 = SuzukiTrotter(order=2)
product_formula_st4 = SuzukiTrotter(order=4)
```

## Lie-Trotter

### Single circuit for time evolution

The circuit used to time evolve the initial state using the Lie-Trotter technique is shown below.

```python
single_step_evolution_gates_lt = PauliEvolutionGate(
    hamiltonian, dt, synthesis=product_formula_lt
)#builds a gate that implements one time step of time evolution 
single_step_evolution_lt = QuantumCircuit(n_qubits)
single_step_evolution_lt.append(
    single_step_evolution_gates_lt, single_step_evolution_lt.qubits
)
 
print(
    f"""
Trotter step with Lie-Trotter
-----------------------------
Depth: {single_step_evolution_lt.decompose(reps=3).depth()}
Gate count: {len(single_step_evolution_lt.decompose(reps=3))}
Nonlocal gate count: {single_step_evolution_lt.decompose(reps=3).num_nonlocal_gates()}
Gate breakdown: {", ".join([f"{k.upper()}: {v}" for k, v in single_step_evolution_lt.decompose(reps=3).count_ops().items()])}
"""
)
single_step_evolution_lt.decompose(reps=3).draw("mpl", fold=-1)
```

### Set the operators to measure

To characterize the dynamics of the transverse field Ising model, the following two observables are considered. The magnetization $\frac{1}{N} \sum_{i=0}^{N-1} \langle Z_i \rangle$ measures the average alignment of spins with the field and indicates how ordered the system is. The mean spin correlation $\frac{1}{N-1} \sum_{i=0}^{N-2} \langle Z_i Z_{i+1} \rangle$ measures how strongly neighbouring spins align during the time evolution.

```python
magnetization = (
    SparsePauliOp.from_sparse_list(
        [("Z", [i], 1.0) for i in range(0, n_qubits)], num_qubits=n_qubits
    )
    / n_qubits
)
correlation = SparsePauliOp.from_sparse_list(
    [("ZZ", [i, i + 1], 1.0) for i in range(0, n_qubits - 1)], num_qubits=n_qubits
) / (n_qubits - 1)
print("magnetization : ", magnetization)
print("correlation : ", correlation)
```

Normalized coefficients $1/6$ and $1/5$.

### Perform time evolution

The real time evolution of the transverse field Ising model and the associated observables are simulated. The state is prepared in $|001100\rangle$ and the time evolution circuit is first applied to this initial state. For each subsequent time step the same Trotterized time evolution gate is applied to the state obtained at the previous step, so after $n$ time steps the circuit has been applied $n$ times and the total evolution time is $t = n \Delta t$. At each time StatevectorEstimator (no measurements are performed and the state is never collapsed) is used to compute the energy, magnetization, and mean spin correlation as expectation values of the form $\langle \psi | \hat{O} | \psi \rangle$.

```python
# Initiate the circuit
evolved_state = QuantumCircuit(initial_circuit.num_qubits)
# Start from the initial spin configuration
evolved_state.append(initial_circuit, evolved_state.qubits)
# Initiate Estimator (V2)
estimator = StatevectorEstimator()
# Set number of shots (dictates precision)
shots = 10000
# Translate the precision required from the number of shots
precision = np.sqrt(1 / shots)
energy_list = []
mag_list = []
corr_list = []
# Estimate expectation values for t=0.0
job = estimator.run(
    [(evolved_state, [hamiltonian, magnetization, correlation])], precision=precision
)
# Get estimated expectation values
evs = job.result()[0].data.evs
energy_list.append(evs[0])
mag_list.append(evs[1])
corr_list.append(evs[2])
# Start time evolution
for n in range(num_timesteps):
    # Expand the circuit to describe delta-t
    evolved_state.append(single_step_evolution_gates_lt, evolved_state.qubits)
    # Estimate expectation values at delta-t
    job = estimator.run(
        [(evolved_state, [hamiltonian, magnetization, correlation])],
        precision=precision,
    )
    # Retrieve results (expectation values)
    evs = job.result()[0].data.evs
    energy_list.append(evs[0])
    mag_list.append(evs[1])
    corr_list.append(evs[2])
# Transform the list of expectation values (at each time step) to arrays
energy_array = np.array(energy_list)
mag_array = np.array(mag_list)
corr_array = np.array(corr_list)
```

The expectation values are plotted against time below.

```python
fig, axes = plt.subplots(3, sharex=True)
times = np.linspace(0, evolution_time, num_timesteps + 1)  # includes initial state
axes[0].plot(
    times,
    energy_array,
    label="First order",
    marker="x",
    c="darkmagenta",
    ls="-",
    lw=0.8,
)
axes[1].plot(
    times, mag_array, label="First order", marker="x", c="darkmagenta", ls="-", lw=0.8
)
axes[2].plot(
    times, corr_array, label="First order", marker="x", c="darkmagenta", ls="-", lw=0.8
)
axes[0].set_ylabel("Energy")
axes[1].set_ylabel("Magnetization")
axes[2].set_ylabel("Mean spin correlation")
axes[2].set_xlabel("Time")
fig.suptitle("Observable evolution")
```

The Hamiltonian is time independent, so the energy should be conserved. As expected, only small numerical fluctuations are visible.

Magnetization starts at $\frac{1}{3}$ because $|001100\rangle$ has four spins with $Z=+1$ and two with $Z=-1$. The oscillation frequency is set by the field magnitude, $2|h| = 2\sqrt{(h\cos\alpha)^2 + (h\sin\alpha)^2} = 2h=2.4$. The $ZZ$ interactions introduce dephasing between spins, which causes the oscillation envelope to shrink over time.

Mean spin correlation starts at $\frac{1}{5}(+1 -1 +1 -1 +1) = \frac{1}{5} = 0.2$, since $|001100\rangle$ gives $Z_i Z_{i+1} = +1,-1,+1,-1,+1$ for the five neighbour pairs. It then oscillates below $0.2$ as the $x$-field changes the $z$-spin configuration while the $ZZ$ interaction tries to preserve the original ordering.

## Second-order Suzuki-Trotter

### Single circuit for time evolution

The second order Suzuki-Trotter formula uses the symmetric decomposition $e^{-i(H_1 + H_2)t} \approx e^{-iH_1t/2} e^{-iH_2t} e^{-iH_1t/2}.$

```python
# Modify the line below (Use PauliEvolutionGate)
single_step_evolution_gates_st2 = PauliEvolutionGate(
    hamiltonian, dt, synthesis=product_formula_st2
)
single_step_evolution_st2 = QuantumCircuit(n_qubits)
single_step_evolution_st2.append(
    single_step_evolution_gates_st2, single_step_evolution_st2.qubits
)
# Let us print some stats
print(
    f"""
Trotter step with second-order Suzuki-Trotter
-----------------------------
Depth: {single_step_evolution_st2.decompose(reps=3).depth()}
Gate count: {len(single_step_evolution_st2.decompose(reps=3))}
Nonlocal gate count: {single_step_evolution_st2.decompose(reps=3).num_nonlocal_gates()}
Gate breakdown: {", ".join([f"{k.upper()}: {v}" for k, v in single_step_evolution_st2.decompose(reps=3).count_ops().items()])}
"""
)
single_step_evolution_st2.decompose(reps=2).draw("mpl", fold=-1)
```

The gates are similar to before, but for the terms that are placed in the 'half step' part, the coefficients are halved. The $ZZ$ coefficient is $-0.2/2 = -0.1$, the $Z$ coefficient is $-0.459/2 \approx -0.23$, and the $X$ coefficient is $-1.109/2 \approx -0.554$. The exception is the $X$ term on qubit $5$, with the full middle step, with coefficient $-1.109.$

```python
# Initiate the circuit
evolved_state = QuantumCircuit(initial_circuit.num_qubits)
# Start from the initial spin configuration
evolved_state.append(initial_circuit, evolved_state.qubits)
# Initiate Estimator (V2)
estimator = StatevectorEstimator()
# Set number of shots
shots = 10000
# Translate the precision required from the number of shots
precision = np.sqrt(1 / shots)
energy_list_st2 = []
mag_list_st2 = []
corr_list_st2 = []
# Estimate expectation values for t=0.0
job = estimator.run(
    [(evolved_state, [hamiltonian, magnetization, correlation])], precision=precision
)
# Get estimated expectation values
evs = job.result()[0].data.evs
energy_list_st2.append(evs[0])
mag_list_st2.append(evs[1])
corr_list_st2.append(evs[2])
# Start time evolution
for n in range(num_timesteps):
    # Expand the circuit to describe delta-t
    evolved_state.append(single_step_evolution_gates_st2, evolved_state.qubits)
    # Estimate expectation values at delta-t
    job = estimator.run(
        [(evolved_state, [hamiltonian, magnetization, correlation])],
        precision=precision,
    )
    # Retrieve results (expectation values)
    evs = job.result()[0].data.evs
    energy_list_st2.append(evs[0])
    mag_list_st2.append(evs[1])
    corr_list_st2.append(evs[2])
# Transform the list of expectation values (at each time step) to arrays
energy_array_st2 = np.array(energy_list_st2)
mag_array_st2 = np.array(mag_list_st2)
corr_array_st2 = np.array(corr_list_st2)
```

```python
axes[0].plot(
    times,
    energy_array_st2,
    label="Second Order",
    marker="x",
    c="limegreen",
    ls="-",
    lw=0.8,
)
axes[1].plot(
    times,
    mag_array_st2,
    label="Second Order",
    marker="x",
    c="limegreen",
    ls="-",
    lw=0.8,
)
axes[2].plot(
    times,
    corr_array_st2,
    label="Second Order",
    marker="x",
    c="limegreen",
    ls="-",
    lw=0.8,
)
 
# Replace the legend
# legend.remove()
legend = fig.legend(
    *axes[0].get_legend_handles_labels(),
    bbox_to_anchor=(1.0, 0.5),
    loc="center left",
    framealpha=0.5,
)
fig
```

## Executing on quantum hardware

The time evolution is now run on quantum hardware using a smaller lattice with $N = 2$ and initial spin configuration $\downarrow\uparrow$. By varying $\alpha$, changes in the wavefunction dynamics can be seen.

```python
n_qubits_2 = 2
dt_2 = 1.6 #time step per evolution step
product_formula = LieTrotter(reps=1)
```

```python
# We prepare an initial state ↓↑ (10).
# Note that Statevector and SparsePauliOp interpret the qubits from right to left
initial_circuit_2 = QuantumCircuit(n_qubits_2)
initial_circuit_2.prepare_state("10")
# Change reps and see the difference when you decompose the circuit
initial_circuit_2.decompose(reps=1).draw("mpl")
```

As a reference, the ideal statevector simulator is used first.

```python
bar_width = 0.1
# initial_state = Statevector.from_label("10")
final_time = 1.6
eps = 1e-5
 
# We create the list of angles in radians, with a small epsilon
# the exactly longitudinal field, which would present no dynamics at all
alphas = np.linspace(-np.pi / 2 + eps, np.pi / 2 - eps, 5)
 
for i, alpha in enumerate(alphas):
    evolved_state_2 = QuantumCircuit(initial_circuit_2.num_qubits)
    evolved_state_2.append(initial_circuit_2, evolved_state_2.qubits)
    hamiltonian_2 = get_hamiltonian(nqubits=2, J=0.2, h=1.0, alpha=alpha)
    single_step_evolution_gates_2 = PauliEvolutionGate(
        hamiltonian_2, dt_2, synthesis=product_formula
    )
    evolved_state_2.append(single_step_evolution_gates_2, evolved_state_2.qubits)
    evolved_state_2 = Statevector(evolved_state_2)
    # Dictionary of probabilities
    amplitudes_dict = evolved_state_2.probabilities_dict()
    labels = list(amplitudes_dict.keys())
    values = list(amplitudes_dict.values())
    # Convert angle to degrees
    alpha_str = f"$\\alpha={int(np.round(alpha * 180 / np.pi))}^\\circ$"
    plt.bar(np.arange(4) + i * bar_width, values, bar_width, label=alpha_str, alpha=0.7)
 
plt.xticks(np.arange(4) + 2 * bar_width, labels)
plt.xlabel("Measurement")
plt.ylabel("Probability")
plt.suptitle(
    f"Measurement probabilities at $t={final_time}$, for various field angles $\\alpha$\n"
    f"Initial state: 10, Linear lattice of size $L=2$"
)
plt.legend()
```

The system is prepared in the spin configuration $\downarrow\uparrow$, corresponding to $|\psi(0)\rangle = |10\rangle$. For a transverse field ($\alpha = 0^\circ$) the Hamiltonian mixes the basis states $|10\rangle$ and $|01\rangle$, so the state oscillates between them. At the chosen time $t = 1.6$ this oscillation is close to a full swap, so a measurement almost always gives $\uparrow\downarrow$. For a longitudinal field ($\alpha = \pm 90^\circ$) the Hamiltonian is diagonal in the $Z$ basis and $|10\rangle$ is an eigenstate, so there is no evolution and the system is measured as it was prepared, $\downarrow\uparrow$. For intermediate angles such as $\alpha = \pm 45^\circ$, the field has both transverse and longitudinal components, so $|10\rangle$ only partially mixes with $|01\rangle$, giving a swap probability of about $67\%$, which is due to the specific Hamiltonian parameters and chosen evolution time. This swap probability cannot be seen in the graph above, even though the same code as in the module was used, which produced the following result.

The evolution is now run on real quantum hardware.

```python
circuit_list = []
for i, alpha in enumerate(alphas):
    evolved_state_2 = QuantumCircuit(initial_circuit_2.num_qubits)
    evolved_state_2.append(initial_circuit_2, evolved_state_2.qubits)
    hamiltonian_2 = get_hamiltonian(nqubits=2, J=0.2, h=1.0, alpha=alpha)
    single_step_evolution_gates_2 = PauliEvolutionGate(
        hamiltonian_2, dt_2, synthesis=product_formula
    )
    evolved_state_2.append(single_step_evolution_gates_2, evolved_state_2.qubits)
    evolved_state_2.measure_all()
    circuit_list.append(evolved_state_2)
```

```python
service = QiskitRuntimeService()
backend = service.least_busy(operational=True, simulator=False)
backend.name
```

```python
pm = generate_preset_pass_manager(backend=backend, optimization_level=3)
circuit_isa = pm.run(circuit_list)
```

```python
circuit_isa[1].draw("mpl", idle_wires=False)
```

All of these gates have been used before in this or other labs, except for the RZZGate, $R_{ZZ}(\theta)=\exp\left(-i\frac{\theta}{2} Z \otimes Z\right)$, which implements the $Z \otimes Z$ interaction.

```python
sampler = SamplerV2(mode=backend)
job = sampler.run(circuit_isa)
job_id = job.job_id()
print("job id:", job_id)
```

```python
results = job.result()
```

```python
list_temp = ["00", "01", "10", "11"]
 
for i, alpha in enumerate(alphas):
    # Dictionary of probabilities
    amplitudes_dict = results[i].data.meas.get_counts()
    values = []
    for str_temp in list_temp:
        values.append(
            amplitudes_dict[str_temp] / 4096.0
        )  # divided by default number of shots
    # Convert angle to degrees
    alpha_str = f"$\\alpha={int(np.round(alpha * 180 / np.pi))}^\\circ$"
    plt.bar(np.arange(4) + i * bar_width, values, bar_width, label=alpha_str, alpha=0.7)
 
plt.xticks(np.arange(4) + 2 * bar_width, labels)
plt.xlabel("Measurement")
plt.ylabel("Probabilities")
plt.suptitle(
    f"Measurement probabilities at $t={final_time}$, for various field angles $\\alpha$\n"
    f"Initial state: 10, Linear lattice of size $L=2$"
)
plt.legend()
```

This is very similar to the results from the statevector simulator, except for the cases $\alpha = \pm 45^\circ$. It also matches the module’s statevector results for all angles.

## Fourth-order Suzuki-Trotter

The fourth-order Suzuki-Trotter is also explained in depth in the [Quantum Simulation notes](https://github.com/AnthonySlawski/Quantum-Circuits/blob/main/Quantum_Simulation_notes.pdf). The circuit for a single time step is shown below.

```python
# Modify the line below (Use PauliEvolutionGate)
single_step_evolution_gates_st4 = PauliEvolutionGate(
    hamiltonian, dt, synthesis=product_formula_st4
)
single_step_evolution_st4 = QuantumCircuit(n_qubits)
single_step_evolution_st4.append(
    single_step_evolution_gates_st4, single_step_evolution_st4.qubits
)
# Let us print some stats 
#fourth not second!
print(
    f"""
Trotter step with fourth-order Suzuki-Trotter
-----------------------------
Depth: {single_step_evolution_st4.decompose(reps=3).depth()}
Gate count: {len(single_step_evolution_st4.decompose(reps=3))}
Nonlocal gate count: {single_step_evolution_st4.decompose(reps=3).num_nonlocal_gates()}
Gate breakdown: {", ".join([f"{k.upper()}: {v}" for k, v in single_step_evolution_st4.decompose(reps=3).count_ops().items()])}
"""
)
single_step_evolution_st4.decompose(reps=2).draw("mpl", fold=-1)
```

The fourth-order Suzuki-Trotter step has a much larger depth and gate count compared with first and second order. The measurement probabilities for the fourth-order case are shown below.

```python
for i, alpha in enumerate(alphas):
    evolved_state_2 = QuantumCircuit(initial_circuit_2.num_qubits)
    evolved_state_2.append(initial_circuit_2, evolved_state_2.qubits)
    hamiltonian_2 = get_hamiltonian(nqubits=2, J=0.2, h=1.0, alpha=alpha)
    single_step_evolution_gates_2 = PauliEvolutionGate(
        hamiltonian_2, dt_2, synthesis=product_formula_st4
    )
    evolved_state_2.append(single_step_evolution_gates_2, evolved_state_2.qubits)
    evolved_state_2 = Statevector(evolved_state_2)
    # Dictionary of probabilities
    amplitudes_dict = evolved_state_2.probabilities_dict()
    labels = list(amplitudes_dict.keys())
    values = list(amplitudes_dict.values())
    # Convert angle to degrees
    alpha_str = f"$\\alpha={int(np.round(alpha * 180 / np.pi))}^\\circ$"
    plt.bar(np.arange(4) + i * bar_width, values, bar_width, label=alpha_str, alpha=0.7)
 
plt.xticks(np.arange(4) + 2 * bar_width, labels)
plt.xlabel("Measurement")
plt.ylabel("Probability")
plt.suptitle(
    f"Measurement probabilities at $t={final_time}$,using fourth-order Suzuki-Trotter\n"
    f"for various field angles $\\alpha$. Initial state: 10, Linear lattice of size $L=2$"
)
plt.legend()
```

This looks very similar to the second-order case, even though it is asymptotically more accurate, however the increased circuit depth makes it far less practical for current quantum hardware. Even on an ideal statevector simulator the differences are negligible for this example.

## Conclusion

In this lab the real time dynamics of the transverse field Ising model were simulated using Trotterization on both statevector backends and real quantum hardware. First, and second order product formulas were implemented, and their results for energy, magnetization, and mean spin correlation were compared with exact values, showing that higher order formulas are more accurate but require much deeper circuits. For a small two site lattice the dependence of measurement probabilities on the field angle was explored and the hardware results were found to be consistent with the ideal statevector simulations.
